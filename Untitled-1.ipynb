```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.rcParams["figure.figsize"] = (12, 6)
```

```python
def intensity_transform(im, breakpoints):
    """
    Apply piecewise-linear intensity transformation.

    im:
        Grayscale image with values in [0,255]

    breakpoints:
        List/array of [input, output] points
    """

    breakpoints = np.array(breakpoints, dtype=np.float32)

    x = breakpoints[:, 0]
    y = breakpoints[:, 1]

    # Create lookup table
    lut = np.interp(np.arange(256), x, y)

    # Apply LUT
    output = cv2.LUT(im, lut.astype(np.uint8))

    return output
```

```python
img = cv2.imread("emma.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Could not find einstein.jpg")
```

```python
breakpoints = [
    [0, 0],
    [50, 50],
    [100, 150],
    [150, 255],
    [255, 255]
]

transformed = intensity_transform(img, breakpoints)
```

```python
bp = np.array(breakpoints)

plt.figure(figsize=(6, 6))
plt.plot(bp[:, 0], bp[:, 1], marker='o')
plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.title("Intensity Transformation")
plt.xlim(0, 255)
plt.ylim(0, 255)
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed, cmap="gray")
plt.title("Transformed")
plt.axis("off")

plt.show()
```

```python
breakpoints2 = [
    [0, 0],
    [70, 30],
    [110, 180],
    [180, 240],
    [255, 255]
]

transformed2 = intensity_transform(img, breakpoints2)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed2, cmap="gray")
plt.title("Modified Intensity Transformation")
plt.axis("off")

plt.show()
```

```python
# Compare several piecewise-linear intensity transformations
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Convert to a single intensity image for a clear transformation comparison.
if img.ndim == 3:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
else:
    gray = img.copy()

gray = np.asarray(gray, dtype=np.uint8)

# Each tuple is (input breakpoint 1, input breakpoint 2,
#                output breakpoint 1, output breakpoint 2).
experiments = [
    (55, 185, 30, 225),
    (70, 175, 35, 235),
    (85, 165, 45, 220),
]

intensities = np.arange(256)
fig, axes = plt.subplots(len(experiments), 3, figsize=(15, 13))
fig.suptitle("Piecewise-Linear Intensity Transformation Experiments", fontsize=16, y=0.995)

for row, (r1, r2, s1, s2) in enumerate(experiments):
    input_points = [0, r1, r2, 255]
    output_points = [0, s1, s2, 255]
    mapping = np.interp(intensities, input_points, output_points).astype(np.uint8)
    result = cv2.LUT(gray, mapping)

    axes[row, 0].plot(intensities, mapping, color="#d95f02", linewidth=2)
    axes[row, 0].plot(input_points, output_points, "o", color="#1b9e77")
    axes[row, 0].plot([0, 255], [0, 255], "--", color="0.65", linewidth=1)
    axes[row, 0].set_title(f"Breakpoints: ({r1}, {r2}) -> ({s1}, {s2})")
    axes[row, 0].set_xlabel("Input intensity")
    axes[row, 0].set_ylabel("Output intensity")
    axes[row, 0].set_xlim(0, 255)
    axes[row, 0].set_ylim(0, 255)
    axes[row, 0].set_aspect("equal")
    axes[row, 0].grid(alpha=0.25)

    axes[row, 1].imshow(gray, cmap="gray", vmin=0, vmax=255)
    axes[row, 1].set_title("Original")
    axes[row, 1].axis("off")

    axes[row, 2].imshow(result, cmap="gray", vmin=0, vmax=255)
    axes[row, 2].set_title("Transformed")
    axes[row, 2].axis("off")

plt.tight_layout(rect=(0, 0, 1, 0.98))
plt.show()
```

```python
brain = cv2.imread("images/q2.jpeg", cv2.IMREAD_GRAYSCALE)



if brain is None:

    raise FileNotFoundError("Could not find images/q2.jpeg")
```

```python
white_breakpoints = [

    [0, 0],

    [80, 0],

    [120, 180],

    [180, 255],

    [255, 255]

]



gray_breakpoints = [

    [0, 0],

    [60, 80],

    [130, 220],

    [200, 255],

    [255, 255]

]



white_matter = intensity_transform(brain, white_breakpoints)
gray_matter = intensity_transform(brain, gray_breakpoints)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(brain, cmap="gray")
plt.title("Original Brain")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(white_matter, cmap="gray")
plt.title("White Matter Enhanced")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(gray_matter, cmap="gray")
plt.title("Gray Matter Enhanced")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
bp = np.array(white_breakpoints)
plt.plot(bp[:, 0], bp[:, 1], marker='o')
plt.title("White Matter Transformation")
plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.grid()

plt.subplot(1, 2, 2)
bp = np.array(gray_breakpoints)
plt.plot(bp[:, 0], bp[:, 1], marker='o')
plt.title("Gray Matter Transformation")
plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.grid()

plt.show()
```

```python
img = cv2.imread("images/WhatsApp Image 2026-09-09 at 20.02.44 (1).jpeg")



if img is None:

    raise FileNotFoundError("Could not find images/WhatsApp Image 2026-09-09 at 20.02.44 (1).jpeg")



img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)



lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)



L, A, B = cv2.split(lab)



print("L range:", L.min(), L.max())
```

```python
gamma = 0.7

L_normalized = L.astype(np.float32) / 255.0

L_gamma = np.power(L_normalized, gamma)

L_corrected = np.uint8(L_gamma * 255)
```

```python
lab_corrected = cv2.merge([L_corrected, A, B])

corrected = cv2.cvtColor(
    lab_corrected,
    cv2.COLOR_LAB2RGB
)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(corrected)
plt.title(f"Gamma Corrected (γ = {gamma})")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(L.flatten(), bins=256, range=(0, 255))
plt.title("Original L Plane Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(L_corrected.flatten(), bins=256, range=(0, 255))
plt.title("Corrected L Plane Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.show()
```
